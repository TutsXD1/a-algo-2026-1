Algoritmo Prim para a distancia demonstrada, tendo como referencia a relacao de distancia ente os polos tecnológicos a seguir:

A -> B = 4KM

A -> C = 4KM

B -> C = 2KM

B -> D = 5KM

C -> D = 5KM

C -> E = 6KM

D -> E = 3KM   (No diagrama diz 4)

D -> F = 4KM

E -> F = 2KM

```python
def algoritmo_prim(grafo, vertice_incial):
    vertices_visitados = {vertice_incial}
    arvore_geradora_minima = []

    while len(vertices_visitados) < len(grafo):
      menor_peso = float('inf')
      menor_aresta = None

      for vertice in vertices_visitados:
        for vertice_adjacente, peso in grafo[vertice]:
          if vertice_adjacente not in vertices_visitados and peso < menor_peso:
            menor_peso = peso
            menor_aresta = (vertice, vertice_adjacente, peso)

      if menor_aresta:
        arvore_geradora_minima.append(menor_aresta)
        vertices_visitados.add(menor_aresta[1])

    return arvore_geradora_minima
```

```python
grafo_lista = {
    'A': [('B', 4), ('C', 4)],
    'B': [('A', 4), ('C', 2), ('D', 5)],
    'C': [('A', 4), ('B', 2), ('D', 5), ('E', 6)],
    'D': [('B', 5), ('C', 5), ('E', 3), ('F', 4)],
    'E': [('C', 6), ('D', 3), ('F', 2)],
    'F': [('D', 4), ('E', 2)]
}

agm = algoritmo_prim(grafo_lista, 'A')
print("Arestas Escolhidas:")
for u, v, w in agm:
    print(f"{u} - {v} : {w}")
```
